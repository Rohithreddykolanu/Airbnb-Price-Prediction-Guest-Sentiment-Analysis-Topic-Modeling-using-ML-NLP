### Text Pre-Processing

```python
# Import required libraries for the Text Pre-Processing
import re                                # For regular expression operations (text cleaning and pattern matching)
import nltk                              # Natural Language Toolkit for NLP tasks
from nltk.corpus import stopwords        # To remove common words that add little meaning
from nltk.stem import WordNetLemmatizer  # For reducing words to their base form (lemmatization)
import string                            # For handling punctuation and string operations
```

```python
# Download required NLTK data
nltk.download('stopwords')  # Stopword corpus for removing common words
nltk.download('wordnet')    # Lexical database for lemmatization
nltk.download('omw-1.4')    # Multilingual WordNet data (supports lemmatizer)
nltk.download('punk_tab')   # Tokenizer models for splitting text into words/sentences
```

```python
# Initialize lemmatizer for reducing words to their base form
lemmatizer = WordNetLemmatizer()

# Define text preprocessing function for Airbnb review comments
def preprocess_text(text):
  if pd.isnull(text):
    return ''

  # Remove URLs, special characters, and convert text to lowercase
  text = re.sub(r'http\S+', '', text)
  text = re.sub(r'[^a-zA-Z\s]', '', text)
  text = text.lower()

  # Tokenize text using simple whitespace splitting
  tokens = text.split()

  # Define custom stopwords specific to Airbnb and Chicago review context
  custom_stopwords = {
    'stay', 'stayed', 'staying', 'host', 'hosts', 'place', 'places', 'property', 'apartment', 'apt', 'cub', 'wrigleyville',
    'home', 'house', 'room', 'bedroom', 'bathroom', 'space', 'area', 'location', 'neighborhood', 'city', 'la', 'muy', 'de',
    'great', 'good', 'nice', 'amazing', 'awesome', 'perfect', 'excellent', 'wonderful', 'comfortable', 'clean', 'und',
    'really', 'very', 'quite', 'much', 'lot', 'recommend', 'enjoy', 'love', 'like', 'overall', 'wrigley', 'milan',
    'chicago', 'downtown', 'loop', 'park', 'lincoln', 'riverwalk', 'navy', 'pier', 'millennium', 'cta'
    }

  # Combine standard English stopwords with custom project-specific stopwords
  stop_words = set(stopwords.words('english')).union(custom_stopwords)

  # Lemmatize words and remove stopwords and punctuation
  tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words and word not in string.punctuation]

  # Return cleaned review text
  return ' '.join(tokens)

# Apply text preprocessing to Airbnb review comments
reviews_df['comments'] = reviews_df['comments'].apply(preprocess_text)
```

***`Observations:`***

- The text preprocessing function standardizes Airbnb review comments by removing URLs, special characters, punctuation, and converting all text to lowercase.

- The use of lemmatization helps reduce words to their base form, which improves consistency in text analysis and reduces vocabulary size.

- Standard English stopwords are combined with custom Airbnb and Chicago specific stopwords to remove commonly repeated words that may not add meaningful analytical value.

- The preprocessing step prepares the review text for NLP tasks such as sentiment analysis, topic modeling, zero shot classification, and feature extraction.

- By applying this function to the comments column, the dataset becomes cleaner and more suitable for extracting meaningful guest experience patterns.

### HTML parsing or HTML tag removal

```python
# HTML parsing and entity decoding libraries
from bs4 import BeautifulSoup
import html

# Define function to clean HTML content from review comments
def clean_html(text):

    # Handle missing or non-text values
    if not isinstance(text, str):
        return ''

    # Decode HTML entities and remove HTML tags
    text = html.unescape(text)
    text = BeautifulSoup(text, "html.parser").get_text(separator=' ')

    # Return cleaned text without extra spaces
    return text.strip()

# Apply HTML cleaning to Airbnb review comments
reviews_df['comments'] = reviews_df['comments'].apply(clean_html)
```

***`Observations:`***

- The HTML cleaning function removes HTML tags and decodes HTML entities from the Airbnb review comments.

- The function also handles missing or non text values by replacing them with empty strings, which prevents errors during text processing.

- This step is important because review comments may contain encoded characters or HTML fragments that can affect NLP model performance.

- After applying HTML cleaning, the comments become more readable and consistent for sentiment analysis, topic modeling, and zero shot classification.

- Overall, this preprocessing step improves the quality of the review text before extracting guest experience insights.

```python
# VADER sentiment analysis library
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize VADER sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Define function to extract VADER sentiment scores
def get_vader_scores(text):

  # Calculate sentiment scores for each review comment
  scores = analyzer.polarity_scores(str(text))

  # Return compound, positive, neutral, and negative sentiment scores
  return pd.Series([scores['compound'], scores['pos'], scores['neu'], scores['neg']])
```

```python
# Create a copy of the reviews dataset for VADER sentiment analysis
vader_sentiment_reviews_df = reviews_df.copy()

# Apply VADER sentiment scoring to review comments
vader_sentiment_reviews_df[['vader_compound', 'vader_positive', 'vader_neutral', 'vader_negative']] = vader_sentiment_reviews_df['comments'].apply(get_vader_scores)

# Classify reviews into positive, neutral, or negative sentiment categories
vader_sentiment_reviews_df['vader_sentiment'] = vader_sentiment_reviews_df['vader_compound'].apply(lambda x:
    'POSITIVE' if x >= 0.5
    else 'NEGATIVE' if x <= -0.5
    else 'NEUTRAL'
    )

# Display reviews with VADER sentiment scores and labels
vader_sentiment_reviews_df.head()
```

## ***Zero-Shot Classification***

```python
# Transformer and GPU acceleration libraries
from transformers import pipeline
import torch

# Initialize zero-shot classifier for review theme detection
classifier = pipeline(
    'zero-shot-classification',
    model = 'typeform/distilbert-base-uncased-mnli',
    device = 0 if torch.cuda.is_available() else -1,
    torch_dtype = torch.float16 if torch.cuda.is_available() else None
)
```

```python
# Define guest experience categories for zero-shot classification
categories = [
    'Location',
    'Amenities',
    'Service',
    'Cleanliness',
    'Safety'
]

# Apply zero-shot classification to identify review themes
results = classifier(
    reviews_df['comments'].astype(str).tolist(),
    candidate_labels = categories,
    multi_label = True
)

# Convert zero-shot classification results into a structured dataframe
zero_shot_df = pd.DataFrame([{

    # Store listing ID associated with each review
    'review_id': reviews_df.iloc[i]['listing_id'],

    # Store classification scores for each guest experience category
    **{label: score for label, score in zip(res['labels'], res['scores'])}
} for i, res in enumerate(results)])
```

### Define functions to create plots

```python
# Define function to visualize top keywords for each topic
def plot_top_words(model, feature_names, n_top_words=10, title = None):

    # Get the number of topics from the fitted topic model
    n_topics = model.components_.shape[0]

    # Create subplot layout for topic keyword charts
    fig, axes = plt.subplots(2, (n_topics + 1) // 2, figsize=(15, 8), sharex=False)
    axes = axes.flatten()

    # Plot top weighted words for each topic
    for topic_idx, topic in enumerate(model.components_):
        top_idx = topic.argsort()[-n_top_words:][::-1]
        top_features = [feature_names[i] for i in top_idx]
        weights = topic[top_idx]
        axes[topic_idx].barh(top_features[::-1], weights[::-1])
        axes[topic_idx].set_title(f"Topic {topic_idx+1}")

    # Hide unused subplot areas
    for ax in axes[n_topics:]:
        ax.axis("off")

    # Add overall title and display plot
    fig.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
# Define function to generate word clouds for topic keywords
def plot_word_clouds(model, feature_names, n_top_words=30, max_words=100, title = None):

    # Get the number of topics from the fitted topic model
    n_topics = model.components_.shape[0]

    # Create subplot layout for topic word clouds
    fig, axes = plt.subplots(2, (n_topics + 1) // 2, figsize=(15, 8))
    axes = axes.flatten()

    # Generate a word cloud for each topic
    for topic_idx, topic in enumerate(model.components_):
        top_idx = topic.argsort()[-n_top_words:]
        freqs = {feature_names[i]: float(topic[i]) for i in top_idx}
        wc = WordCloud(
            width=800,
            height=400,
            background_color='white',
            max_words=max_words
            ).generate_from_frequencies(freqs)
        axes[topic_idx].imshow(wc, interpolation="bilinear")
        axes[topic_idx].axis("off")
        axes[topic_idx].set_title(f"Topic {topic_idx+1}")

    # Hide unused subplot areas
    for ax in axes[n_topics:]:
        ax.axis("off")

    # Display topic word cloud plots
    plt.tight_layout()
    plt.show()
```

***`Observations:`***

- The plotting functions were created to make topic modeling results easier to interpret visually.

- plot_top_words() displays the most important keywords for each topic using horizontal bar charts, helping compare keyword weights across topics.

- plot_word_clouds() generates word clouds for each topic, where larger words represent stronger importance within that topic.

- Both functions automatically detect the number of topics from the fitted model, making them reusable for LDA, NMF, and LSA models.

- These visualizations help translate model outputs into readable topic themes, making it easier to understand what each topic represents in Airbnb guest reviews.

### ***Traditional Topic Modeling Approaches***

#### ***Latent Dirichlet Allocation (LDA)***

```python
# Set the number of topics for LDA topic modeling
n_topics = 5

# Initialize LDA model for identifying hidden review topics
lda_model = LatentDirichletAllocation(n_components = n_topics, random_state = 1)

# Fit LDA model using the Bag-of-Words matrix
lda_model.fit(x_bow)

# Extract feature names from the Bag-of-Words vectorizer
terms = bow_vectorizer.get_feature_names_out()

# Create dictionary to store top words for each LDA topic
lda_topics = {}

# Extract the top 10 keywords from each topic
for index, topic in enumerate(lda_model.components_):
  top_words = [terms[i] for i in topic.argsort()[-10:][::-1]]
  lda_topics[f'Topic {index + 1}'] = top_words

# Display the extracted LDA topics and their top keywords
print('LDA Topics:')
for topic, words in lda_topics.items():
  print(f'{topic}: {', '.join(words)}')
```

#### Topic Distance

```python
# PCA library for reducing topic dimensions
from sklearn.decomposition import PCA

# Define function to visualize topic distances using PCA
def plot_topic_distances_pca(model, title = 'Topic Distance (PCA)'):

  # Extract topic-word matrix from the topic model
  top_word_matrix = model.components_

  # Reduce topic-word matrix into two principal components
  pca = PCA(n_components = 2)
  topic_pca = pca.fit_transform(top_word_matrix)

  # Create scatter plot of topics in 2D PCA space
  plt.figure(figsize = (8, 6))
  plt.scatter(topic_pca[:, 0], topic_pca[:, 1], c = 'blue', edgecolor = 'k', s = 100)

  # Add topic labels to each point
  for i in range(len(topic_pca)):
    plt.text(topic_pca[i, 0], topic_pca[i, 1], f'Topic {i + 1}', fontsize = 12)

  # Add chart title and axis labels
  plt.title(title)
  plt.xlabel('PCA Component 1')
  plt.ylabel('PCA Component 2')

  # Display PCA topic distance plot
  plt.show()
```

```python
# Compare topic distances across all topic modeling methods
plot_topic_distances_pca(lda_model, title = 'LDA Topic Distance (PCA)')  # LDA
plot_topic_distances_pca(nmf_model, title = 'NMF Topic Distance (PCA)')  # NMF
plot_topic_distances_pca(lsa_model, title = 'LSA Topic Distance (PCA)')  # LSA
```

***`Observations:`***

- The LDA topic distance plot shows that the five topics are widely separated, especially Topic 4, indicating that LDA identified relatively distinct themes in the review text.

- In the NMF plot, Topics 3, 4, and 5 are positioned close to each other, suggesting some overlap among these topics, while Topics 1 and 2 are clearly separated and represent more unique themes.

- The LSA plot shows Topic 1 as strongly separated from the remaining topics, meaning it captures a dominant theme that is different from the others.

- Topics that are closer together in the PCA plots likely share similar keywords or review patterns, while topics farther apart represent more distinct guest experience themes.

- Overall, the topic distance visualizations help compare how well each method separates review themes, with LDA showing stronger topic separation, NMF showing moderate overlap, and LSA capturing one highly dominant topic with other topics clustered more closely.

#### Handling Outliers of Independent Variables

```python
# Calculate the first and third quartiles for numeric features
Q1 = listings_df[Numeric_Columns].quantile(0.25)
Q3 = listings_df[Numeric_Columns].quantile(0.75)

# Calculate the interquartile range for each numeric feature
IQR = Q3 - Q1

# Define lower and upper limits for detecting outliers using the IQR method
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Calculate the percentage of outliers in each numeric feature
Outliers_Percentage = (((listings_df[Numeric_Columns] < lower_bound) | (listings_df[Numeric_Columns] > upper_bound)).sum() / listings_df.shape[0]) * 100

# Display outlier percentage for all numeric columns
print(f'Percentage of Outliers in the Numerical Columns: \n{Outliers_Percentage}')
```

***`Observations:`***

- The IQR method was used to detect outliers in the numerical columns by calculating the first quartile, third quartile, interquartile range, lower bound, and upper bound for each feature.

- Several review-based NLP features show a high percentage of outliers, especially Service with 22.11, Safety with 21.90, Location with 19.51, Cleanliness with 19.45, Amenities with 17.63, and huggingface_score with 14.82. This is expected because most review scores are concentrated near high values, while lower scores appear as unusual cases.

- Among the topic modeling features, topic_2 has 21.12 outliers, topic_5 has 15.16 outliers, and topic_3 has 8.46 outliers, indicating that some topics are dominant only for a smaller group of listings. Topic_1 and topic_4 show 0 outliers, suggesting more stable distributions based on the IQR rule.

- Host and review-related features also contain noticeable outliers. host_acceptance_rate has 12.86 outliers, number_of_reviews has 6.82 outliers, and review_scores_rating has 5.13 outliers, showing that some listings have unusually low acceptance rates, very high review counts, or unusually low ratings.

- Property-related features have fewer outliers compared to NLP and review features. accommodates has 3.98 outliers, bathrooms has 1.90 outliers, and bedrooms has only 0.47 outliers, meaning most listings fall within a typical property size range.

- availability_365 has 0 outliers, which means the availability values are spread within the expected range and do not violate the IQR-based limits.

- The target variable price has 6.67 outliers, confirming the presence of unusually high-priced listings compared to the majority of properties. These extreme values may strongly influence model training and prediction errors.

- Since the dataset contains outliers across price, review scores, NLP scores, and topic features, appropriate outlier treatment such as capping or transformation can help reduce the effect of extreme values before building the prediction model.

```python
# Apply IQR based capping also known as Winsorization to selected columns
for col in ['host_acceptance_rate', 'review_scores_rating', 'number_of_reviews']:

    # Calculate the first quartile which represents the 25th percentile
    Q1 = listings_df[col].quantile(0.25)

    # Calculate the third quartile which represents the 75th percentile
    Q3 = listings_df[col].quantile(0.75)

    # Compute the Interquartile Range which measures spread of the middle 50 percent of data
    IQR = Q3 - Q1

    # Define the lower bound and upper bound for acceptable values based on IQR rule
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap values below lower bound and above upper bound to limit the effect of outliers
    listings_df[col] = np.clip(listings_df[col], lower_bound, upper_bound)
```

***`Observations:`***

- IQR based outlier treatment was applied only to selected numerical columns: host_acceptance_rate, review_scores_rating, and number_of_reviews.

- These columns were selected because they showed noticeable outliers during the earlier outlier percentage check and boxplot analysis.

- The method calculates Q1, Q3, and IQR for each selected feature to define acceptable lower and upper bounds.

- Values below the lower bound and above the upper bound were capped instead of removed, which helps preserve the number of records in the dataset.

- Winsorization reduces the influence of extreme values while keeping the original data structure intact.

- This step is useful before model building because extreme values in features like review counts and ratings can disproportionately affect model learning.

- The target variable price was not capped in this step, so the outlier handling was focused only on selected independent variables.

### ***Building Neural Network Regression Models using Keras Sequential***

#### ***Neural Networks Model 1***

```python
# Build the first neural network regression model
nn_model1 = Sequential([

    # Define input layer based on the number of training features
    Input(shape = (X_train.shape[1], )),

    # Add the normalization layer
    normalization_layer,

    # Add hidden dense layers with ReLU activation
    Dense(128, activation = 'relu'),
    Dense(64, activation = 'relu'),
    Dense(32, activation = 'relu'),

    # Output layer for predicting Airbnb listing price
    Dense(1)
])
```

***`Observations:`***

- The neural network is designed as a regression model to predict Airbnb listing prices, with the input layer dynamically matching the number of selected features (23). This ensures the model can effectively utilize all engineered features, including listing attributes, review metrics, sentiment scores, and topic-based features.

- A normalization layer is included at the beginning of the model to standardize input features before training. This is useful when features have different scales; however, since the dataset was already normalized during preprocessing, this step is redundant and may result in double normalization.

- The model architecture consists of three fully connected hidden layers with 128, 64, and 32 neurons using ReLU activation. This layered structure allows the model to capture non-linear relationships between features such as amenities, reviews, and perceived experience, which are important drivers of pricing.

- The final output layer contains a single neuron with no activation function, which is appropriate for a regression task where the goal is to predict a continuous variable (price).

- Overall, the model is structured to learn complex interactions between structured features and unstructured insights (from topic modeling and sentiment analysis), making it suitable for capturing real-world Airbnb pricing dynamics.

```python
# Compile the neural network regression model
nn_model1.compile(
    optimizer = 'adam',
    loss = 'mse',
    metrics = ['mae', 'mse']
)

# Display neural network model architecture
nn_model1.summary()
```

***`Observations:`***

- The model is compiled using the Adam optimizer with Mean Squared Error as the loss function, while both MAE and MSE are tracked as evaluation metrics, making it well suited for a regression task such as predicting Airbnb listing prices in Chicago.

- The architecture follows a funnel-shaped, fully connected feedforward design with a Normalization layer at the input, three progressively shrinking hidden layers, and a single-node output layer, reflecting a structured dimensionality reduction approach.

- The Normalization layer accepts 23 input features and holds 47 non-trainable parameters representing the mean and variance of each feature, ensuring all inputs are scaled to a comparable range before being passed into the dense layers for stable training.

- The first hidden layer expands the input into 128 neurons with 3,072 parameters, the second compresses it to 64 neurons with 8,256 parameters, and the third further reduces it to 32 neurons with 2,080 parameters, forming a progressive feature abstraction path.

- The output layer produces a single continuous value with 33 parameters, which is fully consistent with the regression objective of generating one predicted price value for each Airbnb listing based on the structured property and review-derived input features.

- The total parameter count of 13,488, of which 13,441 are trainable, makes this a relatively lightweight model that reduces the risk of overfitting, which is particularly beneficial given the tabular and moderately sized nature of the Airbnb listings dataset.

***`Observations:`***

- The architecture diagram visually confirms the sequential flow of the neural network, beginning with a Normalization layer and passing through four Dense layers in a top-down manner, with each layer's output serving as the input to the subsequent layer.

- The third Dense layer further reduces the 64-dimensional input to 32 neurons, continuing the funnel-shaped compression strategy and distilling the most predictive signals from the earlier layers into a concise intermediate representation.

- The final Dense layer takes the 32-dimensional input and produces a single output value, which corresponds directly to the predicted listing price, confirming that the network is designed as a regression model with a scalar continuous output.

```python
# Evaluate the first neural network model on the training set
train_loss, train_mae, train_mse = nn_model1.evaluate(X_train, y_train)

# Generate price predictions for the training data
y_pred = nn_model1.predict(X_train).flatten()

# Calculate training performance metrics
train_r2 = r2_score(y_train, y_pred)
train_mape = np.mean(np.abs((y_train - y_pred) / y_train)) * 100
train_rmse = np.sqrt(train_mse)

# Get sample size and number of predictors for adjusted R-squared
n = X_train.shape[0]
k = X_train.shape[1]

# Calculate adjusted R-squared for training data
train_adj_r2 = 1 - ((1 - train_r2) * (n - 1) / (n - k -1))

# Store training performance metrics in a dataframe
nn_model1_Training_Performance = pd.DataFrame({
    'MAE': train_mae,
    'MAPE': train_mape,
    'MSE': train_mse,
    'RMSE': train_rmse,
    'R2 Score': train_r2,
    'Adjusted R2 Score': train_adj_r2
}, index = [0])

# Display training performance results
nn_model1_Training_Performance
```

```python
# Define function to calculate Mean Absolute Percentage Error
def mape_score(targets, predictions):
  return np.mean(np.abs((targets - predictions)/targets)) * 100

# Define function to calculate adjusted R-squared
def adj_r2_score(predictors, targets, predictions):
  r2 = r2_score(targets, predictions)
  n = predictors.shape[0]
  k = predictors.shape[1]

  return 1 - ((1 - r2) * (n - 1) / (n - k -1))

# Define function to evaluate regression model performance
def model_performance_evaluation(model, predictors, targets):

  # Generate predictions using the trained model
  predictions = model.predict(predictors)

  # Calculate regression performance metrics
  mse = mean_squared_error(targets, predictions)
  rmse = np.sqrt(mean_squared_error(targets, predictions))
  mae = mean_absolute_error(targets, predictions)
  mape = mape_score(targets, predictions)
  r2 = r2_score(targets, predictions)
  adj_r2 = adj_r2_score(predictors, targets, predictions)

  # Store performance metrics in a dataframe
  df_perf = pd.DataFrame({
    'MAE': mae,
    'MAPE': mape,
    'MSE': mse,
    'RMSE': rmse,
    'R2 Score': r2,
    'Adjusted R2 Score': adj_r2,
  }, index = [0])

  # Return model performance summary
  return df_perf
```
